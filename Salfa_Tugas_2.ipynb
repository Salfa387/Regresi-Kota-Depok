```python
#import library
import numpy as np
import pandas as pd
import statsmodels.api as sm

#Membuat DataFrame dari data yang diberikan
data = {
    'partisipan': np.arange(1, 12),
    'bm_pendidikan': [264.6, 	157.25, 237.72, 238.64, 75.78, 275.64, 176.31, 256.61, 130.01, 273.88, 147.3],
    'bm_kesehatan': [214.46, 299.85, 62.05, 294.29, 151.73, 267.69, 245.6, 191.75, 234.61, 269.63, 151.04],
    'kemiskinan': [8.07, 11.52, 9.5, 11.16, 16.97, 18.17, 9.47, 15.65, 11.43, 14.23, 5.82]}

df = pd.DataFrame(data)

# Memisahkan variabel independen dan dependen
X = df[['bm_pendidikan', 'bm_kesehatan']]
y = df['kemiskinan']

X = sm.add_constant(X) # Menambahkan kolom konstanta untuk intercept

model = sm.OLS(y, X) #Membangun model regresi
results = model.fit()

print(results.summary())
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go
data = {
    'partisipan': np.arange(1, 12),
    'bm_pendidikan': [264.6, 	157.25, 237.72, 238.64, 75.78, 275.64, 176.31, 256.61, 130.01, 273.88, 147.3],
    'bm_kesehatan': [214.46, 299.85, 62.05, 294.29, 151.73, 267.69, 245.6, 191.75, 234.61, 269.63, 151.04],
    'kemiskinan': [8.07, 11.52, 9.5, 11.16, 16.97, 18.17, 9.47, 15.65, 11.43, 14.23, 5.82]}
df = pd.DataFrame(data)
# Variabel independen
X = df[[ "bm_pendidikan", "bm_kesehatan" ]]
# Variabel dependen
y = df["kemiskinan"]
# 4. MENAMBAHKAN INTERCEPT
X_model = sm.add_constant(X)
# 5. ESTIMASI MODEL OLS
model = sm.OLS(y, X_model)
results = model.fit()
# 6. MENAMPILKAN RINGKASAN HASIL REGRESI
print("\n")
print("=" * 70)
print(" HASIL REGRESI LINEAR BERGANDA")
print("=" * 70)
print(results.summary())
# 7. MENGAMBIL KOEFISIEN REGRESI
intercept = results.params["const"]
coef_pendidikan = results.params["bm_pendidikan"]
coef_kesehatan = results.params[ "bm_kesehatan"]
r_squared = results.rsquared
adjusted_r_squared = results.rsquared_adj
# 8. MEMBUAT PERSAMAAN REGRESI
persamaan = (
f"Ŷ = {intercept:.3f}"
f" + ({coef_pendidikan:.3f}) X1"
f" + ({coef_kesehatan:.3f}) X2")
print("\n")
print("=" * 70)
print("PERSAMAAN REGRESI")
print("=" * 70)
print(persamaan)
print(f"\nR2 = {r_squared:.4f}")
print(f"Adjusted R2 = {adjusted_r_squared:.4f}")

# 9. MENGHITUNG NILAI PREDIKSI
df["Prediksi"] = results.predict(X_model)
print("\n")
print("=" * 70)
print(" DATA AKTUAL & PREDIKSI")
print("=" * 70)
print(
    df[[ "partisipan", "bm_pendidikan", "bm_kesehatan", "kemiskinan",
    ]].round(2))
# 10. MEMBUAT GRID UNTUK BIDANG REGRESI
x1 = np.linspace(
df["bm_pendidikan"].min(),
df["bm_pendidikan"].max(),
40)
x2 = np.linspace(
df["bm_kesehatan"].min(),
df["bm_kesehatan"].max(),
40)
X1, X2 = np.meshgrid(x1, x2)

# 11. MENGHITUNG NILAI Y PADA BIDANG REGRESI
Y = ( intercept + coef_pendidikan * X1 + coef_kesehatan * X2)
# 12. MEMBUAT FIGURE 3D
fig = go.Figure()
# 13. MENAMPILKAN DATA AKTUAL
fig.add_trace(
go.Scatter3d(
x=df["bm_pendidikan"],
y=df["bm_kesehatan"],
z=df["kemiskinan"],
mode="markers",
marker=dict(
size=7,
color="red",
opacity=0.9
),
name="Data Aktual",
text=[
f"partisipan {p}"
for p in df["partisipan"]
],
hovertemplate=(
"<b>%{text}</b><br>"

"bm_pendidikan: %{x}<br>"
"bm_kesehatan: %{y}<br>"
"kemiskinan: %{z}"
"<extra></extra>"
)
)
)
# 14. MENAMPILKAN BIDANG REGRESI
fig.add_trace(
go.Surface(
x=X1, y=X2, z=Y,
opacity=0.60,
colorscale="Viridis",
name="Regression Plane",
hovertemplate=(
"bm_pendidikan: %{x:.2f}<br>"
"bm_kesehatan: %{y:.2f}<br>"
"kemiskinan: %{z:.2f}"
"<extra></extra>"
)
)
)
# 15. MENGATUR TAMPILAN GRAFIK
fig.update_layout(
title={
"text": (
"REGRESI LINEAR BERGANDA 3D"
"<br>"
f"<sup>{persamaan} | "
f"R2 = {r_squared:.3f}</sup>" ), "x": 0.5 },
scene={
"xaxis": {"title": "bm_pendidikan (X1)" },
"yaxis": {"title": "bm_kesehatan (X2)" },
"zaxis": { "title": "kemiskinan (Y)" } },
width=1000,
height=750,
margin={
"l": 0, "r": 0, "b": 0, "t": 100 },
legend={ "x": 0.02, "y": 0.98 }
)

# 16. MENAMPILKAN GRAFIK
fig.show()
```
